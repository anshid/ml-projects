```python
import torch
import torch.nn as nn
import torch.optim as optim


N = 100000  # number of samples
x = torch.rand(N, 2) * 200 - 100  # numbers in [-100, 100]
y = x.sum(dim=1, keepdim=True)    # y = x1 + x2

print(x)
print(y)


class AddModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2, 1)  # input size 2, output size 1

    def forward(self, x):
        return self.linear(x)

model = AddModel()

criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.000001) # Reduced learning rate


# Training loop
epochs = 5000

for epoch in range(epochs):
    # Forward pass
    y_pred = model(x)

    # Compute loss
    loss = criterion(y_pred, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.6f}")

```

```python
# Inspect learned parameters
weight = model.linear.weight.data
bias = model.linear.bias.data
print("Learned weight:", weight)
print("Learned bias:", bias)

```

```python
# Test the model on some examples
test = torch.tensor([[5.0, 7.0],
                     [-3.0, 10.0],
                     [1000.0, -250.0],
                     [147971,16436931]])

with torch.no_grad():
    preds = model(test)

print("\nTest inputs:")
print(test)
print("Model predictions:")
print(preds)
print("True sums:")
print(test.sum(dim=1, keepdim=True))
```
